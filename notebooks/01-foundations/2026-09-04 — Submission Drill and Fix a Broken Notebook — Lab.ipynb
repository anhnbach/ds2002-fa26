### How to read an error before you fix it

Python's traceback is read **bottom up**. The last line names the exception and the problem; the line above it points at your code.

```
Traceback (most recent call last):
  File "<stdin>", line 2, in <module>
    print(greeting)
NameError: name 'greeting' is not defined
```

Four exceptions cover most of what you will hit this semester:

| Exception | Usually means |
|---|---|
| `NameError` | The variable was never created, or the cell that created it never ran |
| `TypeError` | You mixed types — text where a number belongs is the classic |
| `KeyError` | That column or dictionary key does not exist. Check spelling and case |
| `FileNotFoundError` | The path is wrong for *this* environment |

### Stretch

Write a `check_orders(df)` function that raises a clear error if the frame is missing either expected column, has zero rows, or contains a negative value in `units`. Then call it on `orders`. This is the shape of validation you will want in both projects.

````python
# stretch

def check_orders(df):
    if 'game' not in df.columns:
        raise ValueError('Missing game column')
    if 'units' not in df.columns:
        raise ValueError('Missing units column')
    if len(df) == 0:
        raise ValueError('The table has no rows')
    if df['units'].min() < 0:
        raise ValueError('Units has a negative value')
    print('Good to go')

check_orders(orders)
````

**Explanation:** I wrote a function that checks for the columns, the table having rows, and no units value being negative. Each check returns an error saying what is wrong. The orders dataframe passes the checks.
